```python
df1 = pd.read_csv("../data/raw/sample_stocks.csv")
df2 = pd.read_csv("../data/raw/sample_stock2.csv")
```

```python
df1["Date"] = pd.to_datetime(df1["Date"])
df1 = df1.sort_values("Date")
```

```python
df1.head()
df1.info()
```

```python
df1["MA20"] = df1["Close"].rolling(20).mean()
df1["MA50"] = df1["Close"].rolling(50).mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(df1["Date"], df1["Close"], label="Close Price")
plt.plot(df1["Date"], df1["MA20"], label="MA20")
plt.plot(df1["Date"], df1["MA50"], label="MA50")

plt.title("Stock Price + Moving Averages")
plt.legend()
plt.show()
```

```python
delta = df1["Close"].diff()

gain = (delta.where(delta > 0, 0)).rolling(14).mean()
loss = (-delta.where(delta < 0, 0)).rolling(14).mean()

rs = gain / loss
df1["RSI"] = 100 - (100 / (1 + rs))
```

```python
df1[["Date", "RSI"]].tail()
```

```python
plt.figure(figsize=(12,4))
plt.plot(df1["Date"], df1["RSI"])
plt.axhline(70, color="red", linestyle="--")
plt.axhline(30, color="green", linestyle="--")

plt.title("RSI Indicator")
plt.show()
```

```python
exp1 = df1["Close"].ewm(span=12, adjust=False).mean()
exp2 = df1["Close"].ewm(span=26, adjust=False).mean()

df1["MACD"] = exp1 - exp2
df1["Signal"] = df1["MACD"].ewm(span=9, adjust=False).mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(df1["Date"], df1["MACD"], label="MACD")
plt.plot(df1["Date"], df1["Signal"], label="Signal Line")

plt.title("MACD Indicator")
plt.legend()
plt.show()
```

```python
print("""
RSI shows overbought/oversold conditions.
MACD shows trend momentum and reversals.
Moving averages show long-term trend direction.
""")
```

```python
## Interpretation of Technical Indicators

- RSI shows overbought/oversold conditions  
- MACD shows trend momentum and reversals  
- Moving averages show long-term trend direction  
```
