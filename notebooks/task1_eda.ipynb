```python
import pandas as pd
import random
from datetime import datetime, timedelta

stocks = ["AAPL", "TSLA", "GOOGL", "AMZN", "MSFT"]
publishers = ["Reuters", "Bloomberg", "CNBC", "Yahoo Finance", "MarketWatch"]

headlines = [
    "Apple reports strong quarterly earnings",
    "Tesla stock surges after delivery numbers",
    "Google announces AI breakthrough",
    "Amazon faces regulatory pressure in EU",
    "Microsoft cloud revenue exceeds expectations",
    "Apple faces supply chain issues",
    "Tesla cuts prices to boost demand",
    "Google ad revenue slows down",
    "Amazon expands logistics network",
    "Microsoft acquires AI startup"
]

data = []

start_date = datetime(2024, 1, 1)

for i in range(200):  # create 200 rows
    date = start_date + timedelta(days=random.randint(0, 365))
    
    row = {
        "headline": random.choice(headlines),
        "url": f"https://news.example.com/{i}",
        "publisher": random.choice(publishers),
        "date": date,
        "stock": random.choice(stocks)
    }
    
    data.append(row)

df = pd.DataFrame(data)

df.to_csv("financial_news.csv", index=False)

df.head()
```

```python
def correct_stock(text):
    text = text.lower()

    if "google" in text or "alphabet" in text:
        return "GOOGL"
    elif "amazon" in text:
        return "AMZN"
    elif "tesla" in text:
        return "TSLA"
    elif "apple" in text:
        return "AAPL"
    elif "microsoft" in text:
        return "MSFT"
    else:
        return None

df["stock"] = df["headline"].apply(correct_stock)
df
```

```python
import random
from datetime import datetime, timedelta

stocks = ["AAPL", "TSLA", "GOOGL", "AMZN", "MSFT"]
publishers = ["Reuters", "CNBC", "Yahoo Finance", "MarketWatch", "Bloomberg"]

headlines = [
    "stock rises after strong earnings report",
    "market reacts to tech sector growth",
    "company announces AI innovation",
    "shares fall due to supply chain issues",
    "investors respond to quarterly results",
    "strong revenue boosts confidence",
    "regulatory pressure impacts stock",
    "new product launch drives optimism"
]

new_data = []
start_date = datetime(2024, 1, 1)

for i in range(300):  # 👈 ADD 300 MORE ROWS
    date = start_date + timedelta(days=random.randint(0, 90))
    stock = random.choice(stocks)

    new_data.append([
        random.choice(headlines),
        f"https://finance.yahoo.com/news/{stock.lower()}-{i}",
        random.choice(publishers),
        date,
        stock
    ])

df_new = pd.DataFrame(new_data, columns=df.columns)

df = pd.concat([df, df_new], ignore_index=True)
df.head()
```

```python
import matplotlib.pyplot as plt

news_trend = df.groupby(df["date"].dt.date).size().sort_index()

plt.figure(figsize=(12,5))
plt.plot(news_trend.index, news_trend.values)
plt.title("News Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Articles")
plt.xticks(rotation=45)
plt.show()
```
